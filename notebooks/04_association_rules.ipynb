````python
import pandas as pd
import numpy as np

df_products = pd.read_csv("customer_products_v2.csv")

print("Dataset shape:", df_products.shape)
display(df_products.head())
````

````python
product_columns = [
    "Savings_Account",
    "Credit_Card",
    "Debit_Card",
    "Insurance",
    "Home_Loan",
    "Personal_Loan"
]
````

````python
transactions_df = df_products[product_columns].astype(bool)

print("Transaction matrix shape:", transactions_df.shape)

display(transactions_df.head())
````

````python
transactions_df = transactions_df[
    transactions_df.sum(axis=1) > 0
].copy()

print("Valid customer transactions:", len(transactions_df))
````

````python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(
    transactions_df,
    min_support=0.10,
    use_colnames=True
)

print("Number of frequent itemsets:", len(frequent_itemsets))
````

````python
display(frequent_itemsets.sort_values(
    by="support",
    ascending=False
).reset_index(drop=True))
````

````python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.50
)

print("Number of association rules:", len(rules))
````

````python
rules_display = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

rules_display["antecedents"] = rules_display["antecedents"].apply(
    lambda x: ", ".join(sorted(x))
)

rules_display["consequents"] = rules_display["consequents"].apply(
    lambda x: ", ".join(sorted(x))
)

rules_display = rules_display.sort_values(
    by="lift",
    ascending=False
).reset_index(drop=True)

display(rules_display)
````

````python
recommendation_rules = rules.copy()
recommendation_rules = recommendation_rules[
    recommendation_rules["consequents"].apply(lambda x: len(x) == 1)
].copy()
````

````python
recommendation_rules = recommendation_rules[
    (recommendation_rules["confidence"] >= 0.50) &
    (recommendation_rules["lift"] > 1.0) &
    (recommendation_rules["support"] >= 0.10)
].copy()
````

````python
recommendation_rules = recommendation_rules.sort_values(
    by=["lift", "confidence"],
    ascending=False
).reset_index(drop=True)
print(
    "Number of recommendation rules:",
    len(recommendation_rules)
)
````

````python
recommendation_display = recommendation_rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

recommendation_display["antecedents"] = (
    recommendation_display["antecedents"]
    .apply(lambda x: ", ".join(sorted(x)))
)

recommendation_display["consequents"] = (
    recommendation_display["consequents"]
    .apply(lambda x: ", ".join(sorted(x)))
)

display(recommendation_display)
````

````python
def recommend_products(
    customer_products,
    rules_df,
    top_n=5
):
    
    recommendations = []

    customer_products = set(customer_products)

    for _, rule in rules_df.iterrows():

        antecedent = set(rule["antecedents"])
        consequent = set(rule["consequents"])

        # Check whether customer owns all antecedent products
        if antecedent.issubset(customer_products):

            # Recommend only products the customer doesn't own
            new_products = consequent - customer_products

            for product in new_products:

                recommendations.append({
                    "Recommended_Product": product,
                    "Confidence": rule["confidence"],
                    "Lift": rule["lift"],
                    "Support": rule["support"]
                })

    if not recommendations:
        return pd.DataFrame(
            columns=[
                "Recommended_Product",
                "Confidence",
                "Lift",
                "Support"
            ]
        )

    recommendations_df = pd.DataFrame(recommendations)

    # Remove duplicate recommendations
    recommendations_df = (
        recommendations_df
        .sort_values(
            by=["Lift", "Confidence"],
            ascending=False
        )
        .drop_duplicates(
            subset=["Recommended_Product"]
        )
        .head(top_n)
        .reset_index(drop=True)
    )

    return recommendations_df
````

````python
customer = [
    "Savings_Account",
    "Debit_Card"
]

recommendations = recommend_products(
    customer,
    recommendation_rules
)

display(recommendations)
````

````python
def get_customer_products(row):
    return [
        product
        for product in product_columns
        if row[product]
    ]


customer_recommendations = []

for _, row in df_products.iterrows():

    owned_products = get_customer_products(row)

    recommendations = recommend_products(
        owned_products,
        recommendation_rules,
        top_n=3
    )

    for _, rec in recommendations.iterrows():

        customer_recommendations.append({
            "Customer_ID": row["Customer_ID"],
            "Existing_Products": ", ".join(owned_products),
            "Recommended_Product": rec["Recommended_Product"],
            "Confidence": rec["Confidence"],
            "Lift": rec["Lift"],
            "Support": rec["Support"]
        })

customer_recommendations_df = pd.DataFrame(
    customer_recommendations
)

print(
    "Total recommendations:",
    len(customer_recommendations_df)
)

display(customer_recommendations_df.head(10))
````

````python
customer_recommendations_df.to_csv(
    "customer_cross_sell_recommendations.csv",
    index=False
)

print("Cross-sell recommendations saved successfully!")
````

## Objective

The objective of this phase was to build a banking cross-sell recommendation system using customer product ownership data and association rule mining.

## 1. Product Dataset

A synthetic banking product dataset was created containing:

- Customer_ID
- Savings_Account
- Credit_Card
- Debit_Card
- Insurance
- Home_Loan
- Personal_Loan
