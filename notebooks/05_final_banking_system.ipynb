# PHASE 10 — FINAL BANKING RISK & CROSS-SELL SYSTEM

## Objective

The objective of Phase 10 is to integrate the final credit-risk prediction
model developed in the previous phases with the cross-sell recommendation
system developed using association rules.

The final system will:

1. Load the trained credit-risk model.
2. Load the final feature list.
3. Load customer risk data.
4. Predict the probability of credit-card default.
5. Classify customers based on the selected threshold.
6. Generate a banking risk decision.
7. Integrate cross-sell product recommendations.
8. Generate a final customer-level analysis.
9. Provide an interactive banking decision-support system.

### Final System Flow

Customer Data
      ↓
Feature Preparation
      ↓
Random Forest Model
      ↓
Default Probability
      ↓
Risk Classification
      ↓
Banking Decision
      ↓
Cross-Sell Recommendation
      ↓
Final Customer Analysis

```python
cross_sell_summary = (
    cross_sell_df
    .groupby("Customer_ID")
    .agg(
        Recommended_Products=("Recommended_Product", 
                              lambda x: ", ".join(x)),
        Recommendation_Count=("Recommended_Product", "count"),
        Max_Confidence=("Confidence", "max"),
        Max_Lift=("Lift", "max"),
        Max_Support=("Support", "max")
    )
    .reset_index()
)

print("Cross-sell summary created!")
print("Shape:", cross_sell_summary.shape)

display(cross_sell_summary.head(10))
```

```python
final_customer_df = risk_df.merge(
    cross_sell_summary,
    on="Customer_ID",
    how="left"
)

print("Final customer dataset created!")
print("Shape:", final_customer_df.shape)
```

```python
final_customer_df["Recommended_Products"] = (
    final_customer_df["Recommended_Products"]
    .fillna("No Recommendation")
)

final_customer_df["Recommendation_Count"] = (
    final_customer_df["Recommendation_Count"]
    .fillna(0)
    .astype(int)
)

final_customer_df["Max_Confidence"] = (
    final_customer_df["Max_Confidence"]
    .fillna(0)
)

final_customer_df["Max_Lift"] = (
    final_customer_df["Max_Lift"]
    .fillna(0)
)

final_customer_df["Max_Support"] = (
    final_customer_df["Max_Support"]
    .fillna(0)
)
```

```python
display(
    final_customer_df[
        [
            "Customer_ID",
            "Default_Probability",
            "Risk_Category",
            "Banking_Decision",
            "Recommended_Products",
            "Recommendation_Count",
            "Max_Confidence",
            "Max_Lift",
            "Max_Support"
        ]
    ].head(20)
)
```

```python
def generate_final_action(row):

    if row["Risk_Category"] == "High Risk":
        return "Risk Review Required"

    elif (
        row["Risk_Category"] == "Low Risk"
        and row["Recommendation_Count"] > 0
        and row["Max_Confidence"] >= 0.60
        and row["Max_Lift"] > 1.0
    ):
        return "Eligible for Cross-Sell"

    elif row["Risk_Category"] == "Low Risk":
        return "Standard Banking Service"

    else:
        return "Manual Review"
```

```python
final_customer_df["Final_Action"] = (
    final_customer_df.apply(generate_final_action, axis=1)
)

print("Final banking actions generated successfully!")
```

```python
def risk_explanation(customer_id):

    customer = final_customer_df[
        final_customer_df["Customer_ID"] == customer_id
    ]

    if customer.empty:
        print("Customer ID not found.")
        return

    customer = customer.iloc[0]

    print("=" * 60)
    print("             RISK EXPLANATION")
    print("=" * 60)

    print(f"Customer ID          : {customer_id}")
    print(f"Default Probability  : {customer['Default_Probability']:.2%}")
    print(f"Risk Category        : {customer['Risk_Category']}")
    print()

    print("KEY RISK INDICATORS")
    print("-" * 60)

    print(f"Recent Payment Delay : {customer['Recent_Payment_Delay']}")
    print(f"Recent Average Delay : {customer['Recent_Avg_Delay']:.2f}")
    print(f"Maximum Payment Delay: {customer['Max_Payment_Delay']}")
    print(f"Severe Delay Count   : {customer['Severe_Delay_Count']}")
    print(f"Frequent Delay Flag  : {customer['Frequent_Delay_Flag']}")
    print(f"Severe Delay Flag    : {customer['Severe_Delay_Flag']}")

    print()
    print("INTERPRETATION")
    print("-" * 60)

    reasons = []

    if customer["Recent_Payment_Delay"] > 0:
        reasons.append("Recent payment delay is present.")

    if customer["Recent_Avg_Delay"] > 0:
        reasons.append("Average recent payment delay is positive.")

    if customer["Max_Payment_Delay"] > 0:
        reasons.append("Customer has experienced payment delays.")

    if customer["Severe_Delay_Count"] > 0:
        reasons.append(
            f"{int(customer['Severe_Delay_Count'])} severe payment delay(s) detected."
        )

    if customer["Frequent_Delay_Flag"] == 1:
        reasons.append("Frequent payment delays detected.")

    if customer["Severe_Delay_Flag"] == 1:
        reasons.append("Severe payment behavior detected.")

    if len(reasons) == 0:
        reasons.append("No major engineered payment-delay indicators detected.")

    for i, reason in enumerate(reasons, 1):
        print(f"{i}. {reason}")

    print("=" * 60)
```
