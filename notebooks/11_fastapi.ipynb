```python
import joblib

# Load the saved model
model = joblib.load("final_credit_risk_model.pkl")

# Load the final feature list
selected_features = joblib.load("final_features.pkl")

print("Model loaded successfully!")
print("\nModel:")
print(model)

print("\nNumber of final features:", len(selected_features))

print("\nFinal features:")
print(selected_features)
```

```python
import pandas as pd

# Load the feature-engineered dataset
df = pd.read_csv("credit_default_feature_engineered.csv")

print("Dataset shape:", df.shape)

print("\nChecking required features...")

missing_features = [
    feature for feature in selected_features
    if feature not in df.columns
]

if missing_features:
    print("Missing features:", missing_features)
else:
    print("All 23 required features are present!")

# Take one real customer
test_customer = df[selected_features].iloc[[0]]

print("\nTest customer shape:", test_customer.shape)

# Generate probability
test_probability = model.predict_proba(test_customer)[0][1]

print("\nTest Default Probability:", round(test_probability, 6))
```
