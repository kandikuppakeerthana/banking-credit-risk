# Phase 5 — Feature Engineering

## Objective

The objective of Phase 5 was to transform the cleaned dataset into a more informative dataset by creating meaningful financial, repayment-behavior, and customer-profile features. These engineered features are designed to help machine learning models identify patterns associated with credit card default risk.

## Feature Engineering Performed

### 1. Financial Features

Created features to summarize the customer's six-month financial activity:

* `Avg_Bill_Amount` — Average bill amount across six months.
* `Avg_Payment_Amount` — Average payment amount across six months.
* `Total_Bill_6M` — Total bill amount across six months.
* `Total_Payment_6M` — Total payment amount across six months.
* `Max_Bill_Amount` — Maximum bill amount observed during the six-month period.

### 2. Credit Utilization Features

Created features to measure how heavily customers use their available credit:

* `Credit_Utilization` — Average bill amount divided by credit limit.
* `Max_Bill_to_Limit` — Maximum bill amount relative to the customer's credit limit.
* `High_Utilization_Flag` — Indicates whether credit utilization is above 75%.

These features help capture potential financial pressure caused by high credit usage.

### 3. Payment Behavior Features

Created features to analyze how customers repay their credit:

* `Payment_to_Bill_Ratio` — Compares average payment amount with average bill amount.
* `Avg_Payment_Delay` — Average repayment status across six months.
* `Max_Payment_Delay` — Highest repayment-status value observed.
* `Delayed_Payment_Count` — Number of months with delayed repayment.
* `Severe_Delay_Count` — Number of months with more significant repayment delays.
* `Recent_Payment_Delay` — Most recent repayment behavior.
* `Recent_Avg_Delay` — Average repayment status for recent months.
* `Older_Avg_Delay` — Average repayment status for older months.
* `Payment_Delay_Trend` — Difference between recent and older repayment behavior.

These features provide the model with information about repayment consistency and changes in customer behavior over time.

### 4. Customer Profile Features

Additional categorical features were created from existing demographic and financial information:

* `Age_Category` — Groups customers into:

  * Young
  * Adult
  * Middle_Aged
  * Senior

* `Credit_Limit_Category` — Groups customers into:

  * Low
  * Medium
  * High

These features provide a more interpretable representation of customer characteristics.

### 5. Risk Indicator Features

Created simple binary indicators for potentially risky financial behavior:

* `High_Utilization_Flag`
* `Frequent_Delay_Flag`
* `Severe_Delay_Flag`

These features help represent important risk signals in a simple form that can be easily interpreted.

## Data Quality Validation

After feature engineering, the dataset was checked for:

* Missing values
* Infinite values
* Number of rows and columns
* Correct creation of engineered features
* Data types of newly created columns

Ratio-based features were handled carefully to avoid division-by-zero problems.
